### Codebasics Student Performance Tracker

```python
students = {
    "Aarav":   [85, 90, 78],
    "Priya":   [72, 68, 75],
    "Rohan":   [45, 52, 48],
    "Sneha":   [95, 92, 98],
    "Manish":  [60, 65, 70],
}
```

```python
for name, scores in students.items():
    total = 0

    for score in scores:
        total += score
    avg = total/len(scores)
    if avg>79:
        remark = "Topper"
    elif avg >59:
        remark = "Pass"
    else:
        remark = "Needs Improvement"
    print(f"{name}: {avg:.2f} : {remark}")
```
